````python
import gradio as gr
import time
````

````python
## User Interface Takes 3 Thinks

## Input ## Output ## Function
````

````python
## State = Object that created in the previous Situation and Give it back to you ~ Like State From Agentic AI


## Whatever we say it will save in state basically -
## Thats how we do the streaming


## If giving the it will increse the accuracy and Quality of the audio But Increse the Timing for the Inferencing
````

````python
## Function for transcribing the audio

# def transcribe(audio , state=""):
#     """Speech to Text Function USing the Pipeline that we defined"""
#     time.sleep(2)
#     text = p(audio)["text"]
#     state = state + text + " "
#     return state, state
````

````python
def transcribe(audio, state=""):
    if audio is None:
        return state, state

    print("Audio received:", audio)

    time.sleep(2)

    result = p(audio)
    print("ASR result:", result)

    text = result["text"]

    state = state + text + " "

    return state, state
````

````python
import gradio as gr

state = gr.State("")

demo = gr.Interface(
    fn=transcribe,

    inputs=[
        gr.Audio(
            sources="microphone",
            type="filepath"
        ),
        state
    ],

    outputs=[
        gr.Textbox(label="Transcription"),
        state
    ]
)

demo.launch(share=True)
````

````python
p = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-base",
    device=0 if torch.cuda.is_available() else -1
)
````

````python
def transcribe(audio, state=""):

    if audio is None:
        return state, state

    print("Audio received:", audio)

    # ASR
    result = p(audio)

    print("ASR result:", result)

    text = result["text"]

    # Maintain previous transcript
    state = state + text + " "

    return state, state


state = gr.State("")
````

````python

demo = gr.Interface(
    fn=transcribe,

    inputs=[
        gr.Audio(
            sources="microphone",
            type="filepath"
        ),
        state
    ],

    outputs=[
        gr.Textbox(label="Transcription"),
        state
    ]
)

demo.launch(share=True)
````
